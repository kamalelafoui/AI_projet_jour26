## 3. Matrice de risque (P × I)

```python
import pandas as pd
df = pd.DataFrame([{'Menace':t['n'],'Cible':t['cible'],'P':t['P'],'I':t['I'],'Score':t['P']*t['I'],'Niveau':level(t['P']*t['I'])} for t in threats(DEFAULT_ARCH)]).sort_values('Score',ascending=False)
df
```

```python
import matplotlib.pyplot as plt
col={'Critique':'#B3261E','Élevé':'#E8710A','Moyen':'#C79A00','Faible':'#2E7D32'}
ax=df.plot.barh(x='Menace',y='Score',color=[col[n] for n in df['Niveau']],legend=False,figsize=(9,3.5))
ax.invert_yaxis(); ax.set_title('Score de risque P × I (sur 25)'); plt.tight_layout(); plt.show()
```
